#Task 1: Data Cleaning & Preprocessing — Netflix Dataset
Step 1: Import libraries and load the dataset

```python
# Import the required libraries
import pandas as pd
import numpy as np

# Load the Netflix dataset
data = pd.read_csv("/content/Dataset.csv")

# Display the first five rows
data.head()
```

Step 2: Explore the dataset
Before cleaning the data, examine its structure and contents.

```python
# Check the number of rows and columns
print("Dataset shape:", data.shape)

# Display column names
print("\nColumn names:")
print(data.columns.tolist())

# Check data types and non-null values
print("\nDataset information:")
data.info()

# Generate descriptive statistics
print("\nDescriptive statistics:")
display(data.describe(include="all"))
```

Step 3: Identify missing and duplicate records
3.1 Check for missing values

```python
# Count missing values in each column
missing_values = data.isnull().sum()

print("Missing values in each column:")
print(missing_values)

# Calculate the percentage of missing values
missing_percentage = (data.isnull().sum() / len(data)) * 100

print("\nPercentage of missing values:")
print(missing_percentage.round(2))
```

3.2 Identify duplicate records

```python
# Count duplicate rows
duplicate_count = data.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

# Display duplicate records, if any
duplicates = data[data.duplicated(keep=False)]

display(duplicates.head(10))
```

Step 4: Standardize inconsistent formats
Standardize text fields by removing extra spaces and ensuring consistent capitalization.

```python
# Standardize text columns
for column in data.select_dtypes(include="object").columns:
    data[column] = data[column].str.strip()

# Standardize selected categorical columns
if "type" in data.columns:
    data["type"] = data["type"].str.title()

if "rating" in data.columns:
    data["rating"] = data["rating"].str.upper()

# Display the cleaned categorical values
if "type" in data.columns:
    print("Content types:")
    print(data["type"].value_counts())

if "rating" in data.columns:
    print("\nRatings:")
    print(data["rating"].value_counts().head(10))
```

Step 5: Transform date-related columns
The date_added column may initially be stored as text. Convert it to a proper datetime format so that you can analyze content by month or year.

```python
# Convert date_added to datetime
if "date_added" in data.columns:
    data["date_added"] = pd.to_datetime(
        data["date_added"],
        errors="coerce"
    )

# Extract the year and month content was added
if "date_added" in data.columns:
    data["date_added_year"] = data["date_added"].dt.year
    data["date_added_month"] = data["date_added"].dt.month
    data["date_added_month_name"] = (
        data["date_added"].dt.month_name()
    )

# Display the transformed columns
data[
    [
        col for col in [
            "title",
            "date_added",
            "date_added_year",
            "date_added_month",
            "date_added_month_name"
        ]
        if col in data.columns
    ]
].head()
```

Step 6: Transform categorical columns
Categorical columns describe groups, such as whether a title is a movie or TV show.

Convert appropriate columns to Pandas' categorical data type.

```python
# Convert categorical columns
categorical_columns = ["type", "rating"]

for column in categorical_columns:
    if column in data.columns:
        data[column] = data[column].astype("category")

# Check the updated data types
print(data.dtypes)
```

Step 7: Clean and transform the duration column
The duration column contains values such as "90 min" for movies and "2 Seasons" for TV shows.



```python
# Extract numerical duration
data["duration_value"] = pd.to_numeric(
    data["duration"].str.extract(r"(\d+)")[0],
    errors="coerce"
)

# Extract duration unit
data["duration_unit"] = data["duration"].str.extract(
    r"(min|Seasons?|Episodes?)",
    expand=False
)

# Standardize duration units
data["duration_unit"] = data["duration_unit"].replace({
    "Seasons": "Season",
    "Episodes": "Episode"
})

# Display the results
data[
    [
        col for col in [
            "title",
            "type",
            "duration",
            "duration_value",
            "duration_unit"
        ]
        if col in data.columns
    ]
].head(10)
```

Step 8: Save the cleaned dataset

```python
# Save the cleaned dataset
data.to_csv("netflix_titles_cleaned.csv", index=False)

print("Cleaned dataset saved successfully as netflix_titles_cleaned.csv")
```
